```python
import torch
```

```python
x=torch.tensor([1.,2.,3.,4.])
y_true=torch.tensor([2.,4.,6.,8.])

w=torch.tensor(0.0,requires_grad=True)

lr=0.1
epoch=10
for i in range(epoch):

    #Step 1: Pred
    y_pred=w*x

    # step 2: Loss
    loss=((y_pred-y_true)**2).mean()
    loss.backward()

    # Step 3:

    with torch.no_grad():
        w-=lr*w.grad
    w.grad.zero_()
    print(f"Epoch {i+1} : w={w.tolist()}, loss{loss.item():.4f}")
```
