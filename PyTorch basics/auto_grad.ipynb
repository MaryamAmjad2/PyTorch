```python
import torch
```

```python
w=torch.tensor(1.315,requires_grad=True)
x=torch.tensor(3.0)
y=torch.tensor(4.0)
y_pred=x*w
loss=(y_pred- y)**2
print(loss)
loss.backward()
print(w.grad)

```

```python
w2=torch.tensor(3., requires_grad=True)
x2=torch.tensor(4.)
y_true=torch.tensor(10.)
learn_rate=0.1
y_pred=w2*x2

loss=(y_pred-y_true)**2
loss.backward()

print("BEFORE UPDATE")
print('w :',w2.item())
print("Loss :",loss.item())
print('grad :', w2.grad.item())

with torch.no_grad():              # No tracking
    w2 -= learn_rate*w2.grad
w2.grad.zero_()                    # Reset gradient

print("AFTER UPDATE")
print('w :',w2.item())


```
