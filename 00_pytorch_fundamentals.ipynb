## Matrix Multiplication

```python
M1=torch.tensor([
    [2,3,4],
    [6,7,4]
])

M2=torch.tensor([
    [10,89,19],
    [23,6,7],
    [4,5,6]
])

# element_wise_mul=M1*M2
# print(element_wise_mul)

matrix_mul=torch.matmul(M1,M2)
print("Order is ", M1.shape, M2.shape)
print(matrix_mul)
```

## Dealing with tensor shape error

```python
M3=torch.rand(3,4)
M4=torch.rand(3,4)

# M5=torch.matmul(M3,M4)    will through error

M4_transpose=M4.T
M5=torch.matmul(M3,M4_transpose)
print(M5.shape)
print(M5)

```

# Aggregation

## Squeeze and UnSqueeze

```python
x = torch.zeros(1, 3, 2)
print(x.shape)          # torch.Size([1, 3, 1])

y = x.squeeze()
print(y.shape)           # torch.Size([3])  ← both 1's removed
```

```python
unsqueezed_y=y.unsqueeze(dim=2)
print(unsqueezed_y.shape)
print(y)
```

# Pytorch Tensors and Numpy

## Changing Data from Numpy to Torch Tensor

```python
np_arr=np.array([1.,2.,3.])

# Numpy Arrray default dtype if float64 and PyTorch
# tensor default dtype is float32
tensor_from_np=torch.from_numpy(np_arr).type(torch.float32)
np_arr,tensor_from_np



```

### if we change value of an array the tensor doesnt change

```python
np_arr=np_arr+10
np_arr,tensor_from_np
```

## Tensor to Numpy Array


```python
tensor_to_np_arr=tensor_from_np.numpy().astype(np.float64)
tensor_to_np_arr
```

# Reproducibility
